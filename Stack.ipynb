## Valid Parentheses

```python
class Solution(object):
    def isValid(self, s):
        """
        :type s: str
        :rtype: int
        """
        stack = []
        closeToOpen = {")" : "(", "]": "[", "}": "{"}

        for c in s:
            if c in closeToOpen:
                if stack and stack[-1] == closeToOpen[c]:
                    stack.pop()
                else:
                    return False
            else:
                stack.append(c)

        if not stack:
            return True
        else:
            return False
```

```python
s = Solution()
s.isValid("()[]{}")
```

## Min Stack

```python
class MinStack(object):

    def __init__(self):
        self.stack = []
        self.minStack = []


    def push(self, val):
        """
        :type val: int
        :rtype: None
        """
        self.stack.append(val)
        if self.minStack:
            if self.minStack[-1] > val:
                self.minStack.append(val)
            else:
                self.minStack.append(self.minStack[-1])
        else:
            self.minStack.append(val)

    def pop(self):
        """
        :rtype: None
        """
        self.stack.pop()
        self.minStack.pop()


    def top(self):
        """
        :rtype: int
        """
        return self.stack[-1]


    def getMin(self):
        """
        :rtype: int
        """
        return self.minStack[-1]
```

## Evaluate Reverse Polish Notation

```python
class Solution(object):
    def evalRPN(self, tokens):
        """
        :type tokens: List[str]
        :rtype: int
        """
        stack = []
        operands = ["+", "-", "/", "*"]

        for t in tokens:
            if t in operands:
                b = int(stack.pop())
                a = int(stack.pop())
                if t == "+":
                    stack.append(a + b)
                elif t == "*":
                    stack.append(a * b)
                elif t == "-":
                    stack.append(a - b)
                else:
                    stack.append(int(float(a) / b))
            else:
                stack.append(t)
        return int(stack.pop())

```

```python
s = Solution()
s.evalRPN(["2", "1", "+", "3", "*"])
```

## Generate Parentheses


```python
class Solution(object):
    def generateParenthesis(self, n):
        """
        :type n: int
        :rtype: List[str]
        """
        stack = []
        res = []

        def backtrack(openN, closedN):
            if openN == closedN == n:
                res.append("".join(stack))
                return

            if openN < n:
                stack.append("(")
                backtrack(openN + 1, closedN)
                stack.pop()
            if closedN < openN:
                stack.append(")")
                backtrack(openN, closedN + 1)
                stack.pop()

        backtrack(0, 0)
        return res
```

```python
s = Solution()
s.generateParenthesis(3)
```

## Daily Temperatures

```python
class Solution(object):
    def dailyTemperatures(self, temperatures):
        """
        :type temperatures: List[int]
        :rtype: List[int]
        """
        res = [0] * len(temperatures)
        stack = []

        for i in range(len(temperatures)):
            while stack and temperatures[i] > stack[-1][0]:
                res[stack[-1][1]] = i - stack[-1][1]
                stack.pop()
            stack.append((temperatures[i], i))

        return res
```

```python
s = Solution()
s.dailyTemperatures([73, 74, 75, 71, 69, 72, 76, 73])
```

## Car Fleet

```python
class Solution(object):
    def carFleet(self, target, position, speed):
        """
        :type target: int
        :type position: List[int]
        :type speed: List[int]
        :rtype: int
        """
        pair = [(p, s) for p, s in zip(position, speed)]
        pair.sort(reverse=True)
        stack = []

        for i in range(len(pair)):
            stack.append((pair[i][0], pair[i][1]))
            if len(stack) > 1:
                timeA = float(target - stack[len(stack)-2][0]) / float(stack[len(stack)-2][1])
                timeB = float(target - stack[-1][0]) / float(stack[-1][1])
                if timeB <= timeA:
                    stack.pop()
        return len(stack)
```

```python
s = Solution()
s.carFleet(12, [10, 8, 0, 5, 3], [2, 4, 1, 1, 3])
```
