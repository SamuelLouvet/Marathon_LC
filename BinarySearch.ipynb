## Binary Search

```python
class Solution(object):
    def search(self, nums, target):
        """
        :type nums: List[int]
        :type target: int
        :rtype: int
        """
        lp, rp = 0, len(nums) - 1

        while lp <= rp:
            m = int((lp + rp) / 2)

            if nums[m] < target:
                lp = m + 1
            elif nums[m] > target:
                rp = m - 1
            else:
                return m
        return -1

```

```python
s = Solution()
s.search([-1,0,3,5,9,12], 9)
```

## Search 2D Matrix

```python
class Solution(object):
    def searchMatrix(self, matrix, target):
        """
        :type matrix: List[List[int]]
        :type target: int
        :rtype: bool
        """

        lp, rp = 0, len(matrix) - 1

        while lp <= rp:
            m = int((lp + rp) / 2)

            if target < matrix[m][0]:
                rp = m - 1
            elif matrix[m][len(matrix[m]) - 1] < target:
                lp = m + 1
            else:
                lp2, rp2 = 0, len(matrix[m]) - 1

                while lp2 <= rp2:
                    m2 = int((lp2 + rp2) / 2)

                    if matrix[m][m2] < target:
                        lp2 = m2 + 1
                    elif matrix[m][m2] > target:
                        rp2 = m2 - 1
                    else:
                        return True
                return False

        return False
```

```python
s = Solution()
s.searchMatrix([[1,3,5],[7,9,11],[12,13,14]], 13)
```

## Koko Eating Bananas

```python
import math

class Solution(object):
    def minEatingSpeed(self, piles, h):
        """
        :type piles: List[int]
        :type h: int
        :rtype: int
        """
        l, r = 1, max(piles)
        res = r

        while l <= r:
            k = (l + r) // 2

            totalTime = 0
            for p in piles:
                totalTime += math.ceil(float(p) / k)
            if totalTime <= h:
                res = k
                r = k - 1
            else:
                l = k + 1
        return res
```

```python
s = Solution()
s.minEatingSpeed([3,6,7,11], 8)
```

## Find Minimum in Rotated Sorted Array

```python
class Solution(object):
    def findMin(self, nums):
        """
        :type nums: List[int]
        :rtype: int
        """
        l, r = 0, len(nums) - 1
        while l < r:
            m = (l + r) // 2
            if nums[m] < nums[r]:
                r = m
            else:
                l = m + 1
        return nums[l]

```

```python
s = Solution()
s.findMin([3,4,5,1,2])
```
