## Binary Search

```python
class Solution(object):
    def search(self, nums, target):
        """
        :type nums: List[int]
        :type target: int
        :rtype: int
        """
        lp, rp = 0, len(nums) - 1

        while lp <= rp:
            m = int((lp + rp) / 2)

            if nums[m] < target:
                lp = m + 1
            elif nums[m] > target:
                rp = m - 1
            else:
                return m
        return -1

```

```python
s = Solution()
s.search([-1,0,3,5,9,12], 9)
```

## Search 2D Matrix

```python
class Solution(object):
    def searchMatrix(self, matrix, target):
        """
        :type matrix: List[List[int]]
        :type target: int
        :rtype: bool
        """

        lp, rp = 0, len(matrix) - 1

        while lp <= rp:
            m = int((lp + rp) / 2)

            if target < matrix[m][0]:
                rp = m - 1
            elif matrix[m][len(matrix[m]) - 1] < target:
                lp = m + 1
            else:
                lp2, rp2 = 0, len(matrix[m]) - 1

                while lp2 <= rp2:
                    m2 = int((lp2 + rp2) / 2)

                    if matrix[m][m2] < target:
                        lp2 = m2 + 1
                    elif matrix[m][m2] > target:
                        rp2 = m2 - 1
                    else:
                        return True
                return False

        return False
```

```python
s = Solution()
s.searchMatrix([[1,3,5],[7,9,11],[12,13,14]], 13)
```
